```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
```

### 01 `relplot`

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
sns.relplot(data=tips,
    x="total_bill", y="tip", height=4, aspect=1);
```

```python
sns.relplot(data=tips,
    x="total_bill", y="tip", 
    height=4, aspect=1,
    col="time");
```

```python
sns.relplot(data=tips,
    x="total_bill", y="tip", 
    height=4, aspect=1,
    size="size");
```

```python
sns.relplot(data=tips,
    x="total_bill", y="tip", 
    height=4, aspect=1,
    hue="smoker");
```

```python
sns.relplot(data=tips,
    x="total_bill", y="tip", 
    height=4, aspect=1,
    hue="smoker", style="smoker");
```

### 02 `lmplot`

```python
sns.lmplot(data=tips, x="total_bill", y="tip",
           height=4, aspect=1,
           line_kws={'color': 'red'});
```

```python
sns.lmplot(data=tips, x="total_bill", y="tip",
           height=4, aspect=1,
           col="time",
           line_kws={'color': 'red'});
```

```python
sns.lmplot(data=tips, x="total_bill", y="tip",
           height=4, aspect=1,
           col="time", hue="smoker");
```
